###  Data Cleaning and Preprocessing 

This notebook focuses on cleaning and preparing a dataset for analysis.The main goal is to ensure the data is accurate, consistent, and ready for further exploratory analysis or modelling.

###  Objectives
The dataset was imported from a CSV file and processed to address common data quality issues, including:
- duplicate records
- missing values
- null values
- inconsistent column naming
- formatting inconsistencies


```python
#Load the data
import pandas as pd 

df= pd.read_csv('heart.csv.csv')
df.head()
```

#### Handling Duplicates

Duplicates rows were identified and removed to avoid skewed results and inaccurate analysis.

```python
# Check for duplicated rows
df.duplicated()
```

```python
# How many duplicated rows are there?
df.duplicated().sum()
```

```python
# View the duplicated rows
df[df.duplicated()]
```

```python
# drop the duplicated rows
df = df.drop_duplicates()
```

```python
df.shape
```

#### Handling Missing Values 



```python
df.isnull().sum()
```

###     Filtering rows 

Columns were reviewed to ensure that values were standardized to improve readability and consistency.

```python
# Check for outliers in the "chol" column
df[df["chol"]>400]
```

### Creating labelled axes 

Replaced the names amd labels with the right ones for sex, target, and exang.

```python
# using .map()
df["sex_label"] = df["sex"].map({1: "Male", 0: "Female"})
```

```python
#check the first few rows of the dataframe to see the new column
df[["sex", "sex_label"]].head()
```

```python
#using .map()
df["target_label"] = df["target"].map({0: "Heart Disease", 1: "No Heart Disease"})
df["has_disease"] = 1 - df["target"]
```

```python
#check the first few rows of the dataframe to see the new column
df[["target", "target_label","has_disease"]].head()   
```

```python
#using .map()
df["exang_label"] = df["exang"].map({1: "Yes", 0: "No"})
#check the first few rows of the dataframe to see the new column
df[["exang", "exang_label"]].head()   
```

```python
#new table with the new columns
df_new = df[["sex_label", "target_label", "exang_label"]]
df_new.head()
```

```python
df.head()
```

### Saving 

This cleaning process improved the quality of the dataset by:

removing duplicate records
reducing inconsistencies caused by missing and null values
ensuring consistent and readable column names
preparing the data for reliable analysis

```python
#saving the cleaned data to a new csv file
df.to_csv('heart_cleaned.csv', index=False) 
```
