### Data Visualization 

This notebook focuses on visualizing and comparing the explored data with the cleaned data. The goal is to understand the characteristics of the dataset before and after cleaning, identify patterns, and validate the impact of data preprocessing on the quality of the analysis. 

### Ojectives 

- Visualize trends in data.
- Make 5 charts (countplot, histogram, boxplot, scatterplot, heatmap)
- Use Matplotlib and Seaborn.

```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt 

# Load the dataset
df =pd.read_csv('heart_cleaned.csv')
```

```python
# how many patients have heart disease by sex
sns.countplot(data=df, x="sex_label", hue="target_label")
plt.title("Heart Disease by Sex")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.legend(title="Heart Disease")
plt.show()
```

#### Explanation 

205 men about 114(around 55%) have heart disease. Of about 100 women about 26(around 26%) have heart disease.Men are twice as much to have the disease than women do in this sample.



```python
#histogram of age distribution by heart disease group
sns.histplot(data=df, x="age", hue="target_label", multiple="dodge", bins=20)
plt.xlabel("Age")
plt.ylabel("Count")
plt.title("Age Distribution by Heart Disease Group")
plt.show()
```

#### Explanation 

Age Histogram

There are less people under the age 40 with the disease and no disease.The number of people increases from age 40 with 60 the highest in count of people with heart disease.The groups overlap a lot and there are only a few patients under 40 and over 70.

```python
# Boxplot of cholesterol levels by heart disease group
sns.boxplot(data=df, x="target_label", y="chol")
plt.title("Cholesterol Levels by Heart Disease Group")
plt.xlabel("Heart Disease")
plt.ylabel("Cholesterol Level")
plt.show() 


```

####  Explanation

The medians are about 235 and 245 mg/dl and the boxes overlap.The 564 outlier is no disease group.The difference is small compared with the spread so cholesterol its not a reliable separator here

```python
#max heart rate vs age
sns.scatterplot(data=df, x="age", y="thalach",hue="target_label")
plt.xlabel("Age")
plt.ylabel("Max Heart Rate")
plt.title("Max Heart Rate vs Age by Heart Disease Group")
plt.show()
```

##### Explanation 

Scatter plot for max heart rate vs Age by heart disease group shows a negative correlation by age. Healthy patients mainly sit between about 150 and 190bpm and disease patients between about 100 and 150.The group overlap in the middle and there are a few outliers.



```python
#correlation heatmap of the features
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.xlabel("Features")
plt.ylabel("Features")
plt.title("Correlation Heatmap of Features")
plt.show()
```
