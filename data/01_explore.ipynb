# Heart Disease Risk Analysis
 
**Goal:** Explore what factors are linked to heart disease in 303 patients.

**Data:** UCI Heart Disease dataset (Kaggle).

## Column Descriptions 
-**age:** Patient's age in years

-**sex:** Patient's sex (1= male, 0= female)

-**cp:**  Chest Pain type (0-3), a category, not an amount.

-**trestbps:** Resting blood pressure in mm Hg when admitted.

-**chol:** Serum cholesterol in mg/dl.

-**fbs:** Fasting blood sugar over 120 mg/dl (1 =yes, 0 =no).

-**restecg:** Resting ECG result(0-2).

-**thalach:** Maximum heart rate reached during the exercise test.

-**exang:** Exercise-induced chest pain (1 =yes, 0 = no).

-**oldpeak:** ST depression on the ECG caused by exercise, compared to rest.

-**slope:** Slope of the peak exercise ST segment.

-**ca:** Number of major blood vessels (0-3) colored by fluoroscopy.

-**thal:** Thalassemia blood disorder result, a category.

-**target:** Heart disease (1 = yes, 0 = no).
